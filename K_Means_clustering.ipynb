```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans


X, y_true = make_blobs(
    n_samples=400,
    centers=4,
    n_features=2,
    cluster_std=0.80,
    random_state=42
)


scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


kmeans = KMeans(
    n_clusters=4,
    n_init=10,
    random_state=42
)


y_kmeans = kmeans.fit_predict(X_scaled)


centers = kmeans.cluster_centers_


print("Cluster assignments (first 10):")
print(y_kmeans[:10])

print("\nLocation of cluster centers:")
print(centers)

print("\nNumber of clusters:", kmeans.n_clusters)
print("Inertia:", kmeans.inertia_)


plt.figure(figsize=(10, 7))

plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=y_kmeans,
    cmap='plasma',
    s=60,
    alpha=0.8
)


plt.scatter(
    centers[:, 0],
    centers[:, 1],
    c='black',
    s=250,
    marker='X',
    edgecolors='white',
    linewidths=2,
    label='Centroids'
)

plt.title("K-Means Clustering Algorithm", fontsize=16)
plt.xlabel("Feature 1 (Scaled)")
plt.ylabel("Feature 2 (Scaled)")

plt.legend()
plt.grid(True, linestyle='--', alpha=0.4)
plt.colorbar(label="Cluster Number")

plt.show()
```
